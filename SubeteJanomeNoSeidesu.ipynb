```python
import gensim
model = gensim.models.KeyedVectors.load_word2vec_format('cc.ja.300.vec.gz')
```

```python
import json

from generate_word_json import fetch_lyrics
from similar_words import build_word_candidates

lyrics_list = fetch_lyrics()
print(f"{len(lyrics_list)}件の歌詞を取得しました")

word_candidates = build_word_candidates(lyrics_list, model, max_candidates=20)
print(f"{len(word_candidates)}語分の候補を生成しました")

with open('word.json', 'w', encoding='utf-8') as f:
    json.dump(word_candidates, f, ensure_ascii=False, indent=2)
```
